```python
import json
import random

file_path = "/content/train_v5_ko_game_npc_v2.jsonl"

with open(file_path, "r", encoding="utf-8") as f:
    dataset = [json.loads(line) for line in f]

print("총 데이터 개수:", len(dataset))

samples = random.sample(dataset, 100)

for i, data in enumerate(samples, 1):
    print("\n" + "=" * 80)
    print(f"SAMPLE {i}")
    print("ID:", data.get("id"))
    print(
        f"tone={data.get('tone')} | "
        f"style={data.get('style')} | "
        f"humor={data.get('humor')} | "
        f"relation={data.get('relation')} | "
        f"role={data.get('role')}"
    )
    print("language:", data.get("language"))
    print("\nFACT:")
    print(data.get("fact"))
    print("\nOUT:")
    print(data.get("out"))
```
